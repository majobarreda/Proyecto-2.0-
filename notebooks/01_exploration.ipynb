## Valores faltantes y duplicados

```python
print(redda.isna().sum())
print('Duplicados completos:', redda.duplicated().sum())
```

## Estaciones y parámetros disponibles

```python
from src.utils.lectura import es_texto
for col in redda.columns:
    if es_texto(redda[col]) and redda[col].nunique() <= 60:
        print(col, sorted(redda[col].dropna().astype(str).unique()))
```

## Estadísticas descriptivas por parámetro

Ajustar los nombres de columna si el archivo real usa otros (ver celda anterior).

```python
redda.groupby('id_parameter')['value'].describe()
```

## Distribución del pH

```python
ph = redda[redda['id_parameter'].astype(str).str.upper() == 'PH']
ph['value'].hist(bins=40)
plt.axvline(5.6, color='red', linestyle='--', label='Lluvia ácida (pH < 5.6)')
plt.xlabel('pH'); plt.ylabel('Muestras'); plt.legend(); plt.show()
print('Porcentaje de muestras ácidas:', round((ph['value'] < 5.6).mean() * 100, 1), '%')
```
